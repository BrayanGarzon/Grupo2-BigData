# Parcial — Análisis de entregas con Pandas

La empresa desea analizar el comportamiento de sus entregas y los costos asociados. Revise y complete el código según el enunciado del README.


```python
import pandas as pd

df = pd.read_csv("datos/entregas.csv")

print("====================================")
print("       INFORME DE ENTREGAS")
print("====================================")
print("Registros iniciales:", len(df))
```

```python
print("\nPrimeros registros:")
print(df.head())

print("\nInformación:")
print(df.info())

print("\nValores nulos:")
print(df.isnull().sum())
```

```python
# Preparación de datos

df["distancia_km"] = pd.to_numeric(
    df["distancia_km"],
    errors="coerce"
)

df["costo_base"] = pd.to_numeric(
    df["costo_base"],
    errors="coerce"
)

df["servicio"] = (
    df["servicio"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

```python
# Tarifas por kilómetro

tarifas = {
    "Express": 1800,
    "Estándar": 1200,
    "Programada": 900,
    "Mismo Día": 1500
}

df["tarifa_km"] = 1200

df["costo_total"] = (
    df["costo_base"] +
    df["distancia_km"] * df["tarifa_km"]
)
```

```python
# Análisis general

costo_total = df["costo_total"].sum()

print("\nCosto total de operaciones: $", costo_total)
```

```python
# Análisis por ciudad

costo_ciudad = (
    df.groupby("ciudad")["costo_total"]
    .sum()
)

ciudad_mayor = costo_ciudad.idxmax()

print("Ciudad con mayor costo:", ciudad_mayor)
```
